## 1. Load Raw Dataset

```python
import os
import numpy as np
import pandas as pd

data_path = '../data/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df = pd.read_csv(data_path)
print(f'Raw dataset shape: {df.shape}')
df.head()
```

## 2. Cast `TotalCharges` and Impute New Subscriber Charges

```python
# Step 1: Strip whitespace and coerce blank values to NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].str.strip(), errors='coerce')

# Step 2: Impute 0.0 for new customers with tenure = 0
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)

print(f'Missing TotalCharges post-cleaning: {df["TotalCharges"].isna().sum()}')
print(f'TotalCharges dtype: {df["TotalCharges"].dtype}')
```

## 3. Verify Customer ID Uniqueness and Summary Distributions

```python
duplicate_count = df['customerID'].duplicated().sum()
print(f'Duplicate Customer IDs: {duplicate_count}')

# Summary statistics for numerical attributes
df.describe().round(2)
```

## 4. Export Cleaned Production Dataset

```python
output_path = '../data/telco_customer_churn_clean.csv'
df.to_csv(output_path, index=False)
print(f'Cleaned dataset successfully exported to: {output_path}')
print(f'Cleaned shape: {df.shape}')
```

## 5. Verification Check on Exported Dataset

```python
df_verify = pd.read_csv(output_path)
assert df_verify.shape == (7043, 21), 'Shape mismatch!'
assert df_verify['TotalCharges'].isnull().sum() == 0, 'Null values present!'
assert pd.api.types.is_numeric_dtype(df_verify['TotalCharges']), 'TotalCharges is not numeric!'
print('[SUCCESS] All verification assertions passed!')
```

### Summary of Data Cleaning:
- Successfully converted `TotalCharges` from text string to `float64`.
- Imputed 0.00 for the 11 accounts with `tenure = 0`.
- Cleaned dataset saved to `data/telco_customer_churn_clean.csv`.
